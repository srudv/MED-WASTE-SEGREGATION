# Medical Waste Data Visualization — CPCB 2023
Create seven reusable charts and export PNG/SVG versions to `reports/figures/`.

```python
from pathlib import Path
import sys
ROOT=Path.cwd()
if not (ROOT/'data').exists(): ROOT=ROOT.parent
sys.path.insert(0,str(ROOT))
from src.visualization import load_visualization_data, generate_all_charts
df=load_visualization_data(ROOT/'data/processed/medical_waste_cleaned.csv')
print(df.shape)
df.head()
```
